## The big picture

Imagine simulating marbles poured into a jar. You need two things: a description of every marble (where it is, how big it is), and the rules that change them (gravity pulls, marbles cannot overlap). Our bacteria are the same, except they are rods instead of balls, and they grow and split.

In **Vivarium** a simulation is made of a few pieces:

| piece | what it is | in a bacterial twin |
|---|---|---|
| **State** | everything in the simulation right now | all the cells: where each one is, its angle, length and width |
| **Process** | a small piece of code that does ONE job: it looks at the state and returns what should change | grow and divide cells; push cells apart |
| **Step** | like a process, but it runs right after the processes, every tick | remove cells that leave the chamber; record the state |
| **Wires** | which part of the state a process reads and writes | |
| **Composite** | puts it all together and runs the clock | |
| **Emitter** | the camera: records the state every tick | lets you count the cells in every frame |
